###รวม data จาก น้ำฝนจากทุกไฟล์ของ input

```python
import glob
import os
```

```python
base_path = "/content/drive/MyDrive/BSD_mid/Input - (Operation Daily rain from API_ 2025)"
files = glob.glob(f"{base_path}/2025-*/**/*.csv", recursive=True)
```

```python
dfs = []
for f in files:
    df = pd.read_csv(f)
    df["month"] = os.path.basename(os.path.dirname(f))  # 2025-07, 2025-08
    df["source_file"] = os.path.basename(f)
    dfs.append(df)
df_all = pd.concat(dfs, ignore_index=True)
```

#กรองข้อมูลให้มีแค่สถาณีใน station in the model

### กรองข้อมูลใน `df_all` ให้มีเฉพาะสถานีที่อยู่ใน `Station_in_model`

```python
# กรอง df_all โดยใช้รหัสสถานีที่ได้มา
print(f"จำนวนสถานีก่อนกรอง: {df_all['tele_station_oldcode'].nunique()}")
print(f"Shape of df_all before filtering: {df_all.shape}")
filtered_df_all = df_all[df_all['tele_station_oldcode'].isin(stations_to_keep)]
print(f"จำนวนสถานีหลังกรอง: {filtered_df_all['tele_station_oldcode'].nunique()}")
print(f"Shape of df_all after filtering: {filtered_df_all.shape}")

# แสดงข้อมูล 5 แถวแรกของ DataFrame ที่ถูกกรองแล้ว
print("Filtered df_all head:")
display(filtered_df_all.head())
```

```python
filtered_df_all.isnull().sum()
```

```python
# สร้าง mapping จาก tele_station_oldcode ไปยัง tele_station_id
# โดยใช้แถวที่ tele_station_id ไม่เป็นค่าว่าง
id_mapping = filtered_df_all.dropna(subset=['tele_station_id']).set_index('tele_station_oldcode')['tele_station_id'].to_dict()
# สร้างสำเนาของ DataFrame เพื่อหลีกเลี่ยง SettingWithCopyWarning
processed_df = filtered_df_all.copy()
# เติมค่า tele_station_id ที่หายไปโดยใช้ mapping
processed_df['tele_station_id'] = processed_df.apply(
    lambda row: id_mapping.get(row['tele_station_oldcode'], row['tele_station_id'])
    if pd.isna(row['tele_station_id']) else row['tele_station_id'],
    axis=1
)
# อัปเดต filtered_df_all ด้วย DataFrame ที่ประมวลผลแล้ว
filtered_df_all = processed_df
```

```python
filtered_df_all.isnull().sum()
```

```python
# สร้างสำเนาเพื่อดำเนินการและหลีกเลี่ยงการแก้ไข DataFrame เดิมโดยตรง
df_temp = filtered_df_all.copy()
# ขั้นตอนที่ 1: ลบคอลัมน์ 'station_code' และ 'value'
df_temp = df_temp.drop(columns=['station_code', 'value'], errors='ignore')
# ขั้นตอนที่ 2: เตรียมข้อมูลสำหรับ df_temp (จาก filtered_df_all เดิม)
# แปลง 'measure_datetime' เป็น datetime และดึงเดือนออกมา
df_temp['measure_datetime'] = pd.to_datetime(df_temp['measure_datetime'])
df_temp['month_num'] = df_temp['measure_datetime'].dt.month
# ขั้นตอนที่ 3: เตรียมข้อมูลสำหรับ history_rain_filtered โดยสร้างสำเนาเพื่อหลีกเลี่ยง SettingWithCopyWarning
history_rain_processed = history_rain_filtered.copy()
history_rain_processed['measure_datetime'] = pd.to_datetime(history_rain_processed['measure_datetime'])
history_rain_processed['month_num'] = history_rain_processed['measure_datetime'].dt.month
# ขั้นตอนที่ 4: คำนวณค่ามัธยฐานของ 'data' จาก history_rain_processed แยกตาม 'station_code' และ 'month_num'
median_data_by_station_month = history_rain_processed.groupby(['station_code', 'month_num'])['data'].median().reset_index()
median_data_by_station_month = median_data_by_station_month.rename(columns={'data': 'median_rainfall_value'})
# ขั้นตอนที่ 5: เติมค่าว่างใน 'rainfall_value' ของ df_temp
# รวมข้อมูล median เข้ามาใน df_temp ชั่วคราวเพื่อเติมค่า
df_temp = df_temp.merge(
    median_data_by_station_month,
    left_on=['tele_station_oldcode', 'month_num'],
    right_on=['station_code', 'month_num'],
    how='left'
)
# เติมค่าว่างใน 'rainfall_value' ด้วย 'median_rainfall_value'
df_temp['rainfall_value'] = df_temp['rainfall_value'].fillna(df_temp['median_rainfall_value'])
# ลบคอลัมน์ชั่วคราวที่ใช้ในการ merge และคอลัมน์ 'month_num'
df_temp = df_temp.drop(columns=['month_num', 'station_code', 'median_rainfall_value'], errors='ignore')
# อัปเดต filtered_df_all ด้วย DataFrame ที่ประมวลผลแล้ว
filtered_df_all = df_temp
# แสดงข้อมูล 5 แถวแรกของ DataFrame ที่ถูกประมวลผลแล้ว
display(filtered_df_all.head())
```

###ตอบคำถาม

จากกราฟจะพบว่าปริมาณน้ำฝนไม่ได้มีการแจกแจงแบบปกติ กราฟเบ้ขวามากๆ ดังนั้นเราจึงเลือกที่จะใช้ค่ากลางแทนค่าเฉลี่ย

```python
import pandas as pd
import os
from google.colab import drive

# ตรวจสอบว่า filtered_df_all มีคอลัมน์ measure_datetime หรือไม่
if 'measure_datetime' not in filtered_df_all.columns:
    print("Error: 'measure_datetime' column not found in filtered_df_all. Please ensure it exists.")
else:
    # แปลง 'measure_datetime' เป็น datetime หากยังไม่ได้ทำ
    filtered_df_all['measure_datetime'] = pd.to_datetime(filtered_df_all['measure_datetime'])

    # ดึงเดือนและปี
    filtered_df_all['month'] = filtered_df_all['measure_datetime'].dt.month
    filtered_df_all['year'] = filtered_df_all['measure_datetime'].dt.year

    # คำนวณปริมาณน้ำฝนรวมรายเดือนสำหรับแต่ละสถานี
    monthly_rainfall = filtered_df_all.groupby(['tele_station_oldcode', 'year', 'month'])['rainfall_value'].sum().reset_index()

    # เปลี่ยนชื่อคอลัมน์ตามที่ร้องขอ
    monthly_rainfall = monthly_rainfall.rename(columns={
        'tele_station_oldcode': 'Node',
        'rainfall_value': 'rain'
    })

    # จัดเรียงลำดับคอลัมน์ใหม่ตามที่ร้องขอ: Node, rain, month, year
    monthly_rainfall = monthly_rainfall[['Node', 'rain', 'month', 'year']]

    # ตรวจสอบและสร้างโฟลเดอร์ปลายทางถ้ายังไม่มี
    output_dir = '/content/drive/MyDrive/BSD_mid/Output - (Clean monthly rain - 2025)'
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)

    # กำหนดพาธไฟล์
    output_file_path = os.path.join(output_dir, 'monthly_rainfall_summary.csv')

    # บันทึก DataFrame ลงในไฟล์ CSV
    monthly_rainfall.to_csv(output_file_path, index=False)

    print(f"ไฟล์ 'monthly_rainfall_summary.csv' ถูกบันทึกที่: {output_file_path}")

    # แสดงข้อมูล 5 แถวแรกของ DataFrame ที่ถูกสร้างขึ้น
    print("\nMonthly Rainfall Summary (head):")
    display(monthly_rainfall.head())

    # แสดงลิงก์ Google Drive (ต้องแน่ใจว่าได้ mount drive แล้ว)
    # โครงสร้างลิงก์สำหรับไฟล์ใน Google Drive
    # ลิงก์นี้จะนำไปยังโฟลเดอร์ ไม่ใช่ไฟล์โดยตรง ซึ่งเป็นวิธีที่ปลอดภัยและง่ายที่สุด
    # ผู้ใช้สามารถหาไฟล์จากชื่อได้ในโฟลเดอร์นั้น
    print("\nคุณสามารถเข้าถึงโฟลเดอร์ที่มีไฟล์นี้ได้ที่:")
    print("https://drive.google.com/drive/folders/14vaVcr2w-aNm5mmMBV-VpLRmwgMmrgCk?usp=sharing" + output_dir.split('/')[-1])
```
